# Mining Customer Sentiment from Air Canada App Reviews to Identify Service Pain Points #


## Part 1: Scraping the Dataset 

```python
from google_play_scraper import reviews, Sort
import pandas as pd
```

```python
# Scraping the latest 7500 reviews, there are less than 7500 so we will use them all.

ac_reviews, _ = reviews (
    'com.aircanada',
    lang = 'en',
    country ='ca',
    sort = Sort.NEWEST,
    count = 7500
)
```

```python
# Coverting the scraped data into a data frame

df = pd.DataFrame(ac_reviews)

df = df[['at', 'content', 'score']]
df = df.rename(columns={'at': 'date'})

df.head()
```

```python
df.info() # Checking dateframe for errors
```

```python
# Coverting snapshot to CSV

df.to_csv('air_canada_reviews.csv', index=False)
```
